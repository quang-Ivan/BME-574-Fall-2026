```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize'] = [16, 8]
```

```python
xC = np.array([2, 1])
sig = np.array([2, 0.5])
```

```python
theta = np.pi/3
```

```python
R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta), np.cos(theta)]])
```

```python
R
```

```python
nPoints = 10000
points = np.random.randn(2,nPoints)
```

```python
points
```

```python
X = R @ np.diag(sig) @ points + np.diag(xC) @ np.ones((2,nPoints))
```

```python
X
```

```python
plt.scatter(X[0,:],X[1,:])
```

```python
Xavg = np.mean(X, axis=1)
Xavg
```

```python
B = X - np.tile(Xavg,(nPoints,1)).T
```

```python
plt.scatter(B[0,:],B[1,:])
```

```python
U, S, VT = np.linalg.svd(B/np.sqrt(nPoints-1),full_matrices=0)
```
